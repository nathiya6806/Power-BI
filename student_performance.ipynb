```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_csv('/student-mat.csv', sep=';')

print("Dataset loaded successfully!")
```

```python
# ==========================================
# STEP 5: DATA CLEANING AND EDA
# Student Performance Analysis
# ==========================================

# 1. Dataset Information
print("========== DATASET INFORMATION ==========")
df.info()

# 2. Check Missing Values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())
print("Total Missing Values:", df.isnull().sum().sum())

# 3. Check Duplicate Records
print("\n========== DUPLICATE RECORDS ==========")
print("Total Duplicates:", df.duplicated().sum())

# 4. Statistical Summary
print("\n========== STATISTICAL SUMMARY ==========")
display(df.describe())

# 5. Student Performance Analysis
print("\n========== STUDENT PERFORMANCE ==========")
print("Average Final Grade:", round(df['G3'].mean(), 2))
print("Highest Final Grade:", df['G3'].max())
print("Lowest Final Grade:", df['G3'].min())

# 6. Study Time vs Final Grades
print("\n========== STUDY TIME ANALYSIS ==========")
study_analysis = df.groupby('studytime')['G3'].mean()
print(study_analysis)

print("\n========== STEP 5 COMPLETED ==========")
```

```python
# ==========================================
# STEP 6: DATA VISUALIZATION
# Student Performance Analysis
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Create two charts
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# CHART 1: Student Final Grade Distribution
sns.histplot(
    data=df,
    x='G3',
    bins=15,
    kde=True,
    color='skyblue',
    ax=axes[0]
)

axes[0].set_title('Distribution of Student Final Grades')
axes[0].set_xlabel('Final Grade (G3)')
axes[0].set_ylabel('Number of Students')

# CHART 2: Study Time vs Average Final Grade
study_analysis = df.groupby('studytime')['G3'].mean()

sns.barplot(
    x=study_analysis.index,
    y=study_analysis.values,
    color='orange',
    ax=axes[1]
)

axes[1].set_title('Study Time vs Average Final Grade')
axes[1].set_xlabel('Weekly Study Time Category')
axes[1].set_ylabel('Average Final Grade')

axes[1].set_xticks([0, 1, 2, 3])
axes[1].set_xticklabels([
    'Less than 2 hrs',
    '2 to 5 hrs',
    '5 to 10 hrs',
    'More than 10 hrs'
], rotation=15)

plt.tight_layout()
plt.show()

print("Both visualizations generated successfully!")
```
